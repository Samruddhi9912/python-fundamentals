```python
a = "python"
print(a[1]) #character of index 1
print(a[0:3]) #index 0 to index 2
print(a[-1]) #character of last index
print(a[-4:-1]) #index 2 to index 5
print(a[:3]) #index 0 to index 2 (a[0:3])
print(a[3:]) #index 3 to 6 (a[3:6])
print(a[:]) #prints full string
```
