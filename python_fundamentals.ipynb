```python
n = int(input("Enter a number: "))
if n<1:
  print(f"{n} is not a prime number")
for i in range(2, n):
  if (n%i == 0):
    print(f"{n} is not a prime number")
    break
else:
  print(f"{n} is a prime number")
```

```python
n = int(input("Enter a number: "))

i = 1
sum = 0
while(i<=n):
  sum += i
  i+=1

print(f"Sum of first {n} natural numbers is {sum}")
```

```python
n = int(input("Enter a number: "))
fact = 1
for i in range(1,n+1):
  fact = fact*i
print(f"The factorial of {n} is {fact}")
```
